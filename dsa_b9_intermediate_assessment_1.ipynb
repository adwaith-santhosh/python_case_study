## Challenge 2 : Analyze daily expenses and extract insights.
User enters:

Number of transactions
Category and amount

Food 200
Travel 500
Food 300
Shopping 700

### Output
* Total expense
* Highest expense category
* Category-wise total

```python
def analyze_expenses():
    n = int(input("Enter number of transactions: "))

    expenses = {}

    for _ in range(n):
        category, amount = input().split()
        amount = float(amount)

        if category in expenses:  # TODO: store category-wise total
          expenses[category]+=amount
        else:
          expenses[category]=amount

    total=0 # TODO: calculate total expense
    for amount in expenses.values():
      total+=amount

    highest_amount=0 # TODO: find highest category
    for category in expenses:
      if expenses[category]>highest_amount:
        highest_amount=expenses[category]
        highest_category=category


    print("Total Expense:", total)
    print("Highest Category:", highest_category)
    print("Category-wise:", expenses)

analyze_expenses()
```
