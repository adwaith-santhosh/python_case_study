1. Bank Account Manager

a. Simulate a bank account with deposit, withdrawal, and balance
enquiry operations

b. Use variables and data types to store account information

c. Implement conditional statements to check for sufficient balance
before withdrawal

d. Use functions to perform deposit, withdrawal, and balance inquiry
operations.


```python
name=input("enter account holder name: ")
balance=int(input("enter the account balance: "))
def deposit(amount):
  global balance
  balance+=amount
  print(f"Name:{name} Deposited:{amount:.2f} Balance:{balance:.2f}:")
def withdraw(amount):
  global balance
  if(balance>=amount):
    balance-=amount
    print(f"Name:{name} Withdrawn:{amount:.2f} Balance:{balance:.2f}:")
  else:
    print("Insuffient Account Balance !!!")

deposit(1000)
withdraw(3000)
```

2. Weather Data Analyzer

a. Analyze weather data (temperature, humidity, etc.) for a given region

b. Use variables and data types to store weather data

c. Implement conditional statements to determine weather conditions
(sunny, rainy, etc.)

d. Use functions to calculate average temperature and humidity, and print
results

```python
temperature=[26,38,18,34,26]
humidity=[40,56,33,51,49]

def avg_temp():
  avg_t=sum(temperature)/len(temperature)
  print(f"Averarge Temperature:{avg_t:.2f}")
def avg_hum():
  avg_h=sum(humidity)/len(humidity)
  print(f"Average Humidity:{avg_h:.2f}")

avg_temp()
avg_hum()

i=int(input("enter the day in which weather conditions is to be determined:"))
if temperature[i-1]>40 and humidity[i-1]>50:
  print("Sunny")
elif temperature[i-1]>25 and humidity[i-1]>35:
  print("partially sunny")
elif temperature[i-1]>25  and humidity[i-1]>45:
  print("rainy")
elif temperature[i-1]<20 and humidity[i-1]<35:
  print("cold")
```
